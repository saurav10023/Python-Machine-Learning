```python
#create table 
import sqlite3 as sq
connect = sq.connect('example.db')
connect
cursor = connect.cursor()
cursor.execute('''
create table if not exists employees(
                id integer primary key ,
                name text not null , 
                age integer ,
                department text
    )
'''
)

connect.commit()
```

```python
##insert data in then table 
cursor.execute(
    '''
insert into employees(name , age , department)
values('saurav' , 20 , 'cse')

 '''
)

cursor.execute(
    '''
insert into employees(name , age , department)
values( 'bob', 20 , 'cse')

 '''
)
cursor.execute(
    '''
insert into employees(name , age , department)
values('sonu' , 19 , 'aiml')

 '''
)
connect.commit()
```

```python
cursor.execute( 
    '''
    select * from employees

'''
)
rows = cursor.fetchall()
## prnt quaried data 
for row in rows :
    print(row)
```

```python
cursor.execute('''
    delete from employees 
        where name = 'bob'
''')
connect.commit()
```
