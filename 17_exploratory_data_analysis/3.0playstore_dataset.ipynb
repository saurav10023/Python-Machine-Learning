```python
df_copy = df_copy.drop_duplicates(subset=['App'] , keep='first')
```

## EDA

```python
## exploratory data analysis 

numeric_features = [ feature for feature in df_copy.columns if df[feature].dtype != 'O']
categorical_features = [ feature for feature in df_copy.columns if df[feature].dtype == 'O']

print ("we have {} numerical features :  {}".format (len(numeric_features) , numeric_features))
print ("we have {} categorical features :  {}".format (len(categorical_features) ,categorical_features))

```

```python
plt.Figure(figsize=(15,15))
plt.suptitle('ananlysis of numerical features' , fontsize=20 , fontweight = 'bold' , alpha = 0.8 , y=1 )


for i in range(0 , len(numeric_features)):
    sns.kdeplot(x=df[numeric_features[i]] , shade= True , color='r')
    plt.xlabel(numeric_features[i])
    plt.tight_layout()
```
