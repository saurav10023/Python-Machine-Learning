## Polynomial regression Implementation

```python
import numpy as np  
import pandas as pd 
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
x = 6*np.random.rand(100,1) -3 
y = 0.5 * x**2 + 1.5*x +2 + np.random.randn(100,1)

## quadratic equation used - y=0.5x^2 + 1.5X+2+outliers
plt.scatter(x,y,color='r')
plt.xlabel('X dataset')
plt.ylabel('Y dataset')
plt.show()
```

```python
from sklearn.model_selection import train_test_split 
x_train , x_test , y_train , y_test = train_test_split(x,y, test_size=0.2 , random_state=42)
```

```python
### implement a simple linear regression  

from sklearn.linear_model import LinearRegression 
regression_1 = LinearRegression()
```

```python
regression_1.fit(x_train , y_train)
```

```python
from sklearn.metrics import r2_score 
r2_score1 = r2_score(y_test , regression_1.predict(x_test))
r2_score1
```

```python
plt.plot(x_train , regression_1.predict(x_train) , color='r')
plt.scatter(x_train , y_train)
plt.xlabel("X Data Set")
plt.ylabel("Y")
```

## Now PolyNomial Regression 

```python
from sklearn.preprocessing import PolynomialFeatures 
poly=PolynomialFeatures(degree=2 , include_bias=True) 
```

```python

x_train_poly = poly.fit_transform(x_train)
x_test_poly = poly.transform(x_test)
```

```python
x_train_poly
```

```python
x_test_poly
```

```python
from sklearn.metrics import r2_score 
regression_poly = LinearRegression()
regression_poly.fit(x_train_poly , y_train)
y_pred2 = regression_poly.predict(x_test_poly)
score2 = r2_score(y_test , y_pred2)
print(score2) 
```

```python
print(regression_poly.coef_)
```

```python
plt.scatter(x_train , regression_poly.predict(x_train_poly))
plt.scatter(x_train , y_train , color='r')
```

```python
## 3 prediction of new data set 

x_new = np.linspace(-3, 3, 200).reshape(200,1)
x_new_poly = poly.transform(x_new)
x_new_poly
```

```python
y_new = regression_poly.predict(x_new_poly)
# Polynomial regression curve
plt.plot(x_new, y_new, "r-", linewidth=3, label="Polynomial Regression")

# Training data
plt.scatter(x_train, y_train, color="blue", s=30, label="Training points")

# Testing data
plt.scatter(x_test, y_test, color="green", s=30, label="Testing points")

plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.show()
```
