## Understanding Exceptions
handling errors without stopping the program.

Eqcetions are events that disrupts the normal flow of program , occur when error is encountered during program execution    

```python
## Exception try , except block

try: 
    a = b 
except :
    print("The variable is not defined")    
```

```python

try: 
    a = b 
except NameError as ex:
    print(ex) 
```

```python

try: 
    a = 10/0
except ZeroDivisionError as zo:
    print(zo) 
    print("divisor greater than zero")
```

```python
## try except and else block

try :
    num = int(input("enter a number :"))
    result = 10/num 
except ValueError :
    print("that is not a valid number")
except ZeroDivisionError :
    print("you cant' divide by zero")
except Exception as ex:
    print(ex)
else :
    print(f"the result is {result}")            
    
```

```python
## try , except , else , finally
try :
    num = int(input("enter a number :"))
    result = 10/num 
except ValueError :
    print("that is not a valid number")
except ZeroDivisionError :
    print("you cant' divide by zero")
except Exception as ex:
    print(ex)
else :
    print(f"the result is {result}")  
finally :  
    print("Execution completed !!")    
```

# try -- we try 
## except -- if error comes  it executes
## else -- if error not comes it executes
## finally -- if error comes or not it executes
