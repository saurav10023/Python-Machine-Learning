## file operations - read and write files

```python
## read a whole File 

with open ('copied.txt', 'r') as file  :
    content = file.read() ## we are reading the entire content of the file 
    
    print(content)
```

```python
with open ('copied.txt', 'r') as file  :
    for line in file :
        ##print(line) ## we are reading the  content line by line 
        ## we get a extra space between lines
        print(line.strip()) ## using strip extra line gets removed
```

```python
## writing a file (overwriting) removing the previous content we use 'w' write mode
with open('copied.txt' , 'w') as file :
    file.write('hello world?? \n')
    file.write('thhis is the new line')
```

```python
## writing without overwriting we use 'a' append  mode
with open('copied.txt' , 'w') as file :
    file.write('Append operation done')
```

```python
## writing list of lines 

lines = ['first line\n' , 'second line \n' , 'third line \n']

with open('copied.txt' , 'a') as file :
    file.writelines(lines)
```

```python
## binary files (in this we update bytes)

data = b'\x00\x01\x02\x03\x04'
with open('requirment.bin' , 'wb') as file :
    file.write(data)
```

```python
## reading a source fine and writing content to a destination file

with open('requirment.txt' , 'r') as source :
    content = source.read()
with open('destinstion.txt' , 'w') as destination :
    destination.write(content)    
```

```python
## counting lines , words and characters in text file 

def count_text(file_path):
    with open (file_path , 'r') as file :
        lines = file.readlines()
        line_count = len(lines) 
        word_count = sum(len(line.split()) for line in lines)
        char_count = sum(len(line) for line in lines)
        return line_count , word_count ,char_count

file_path = 'copied.txt'
lines , words , characters = count_text(file_path)
print( f'Lines : { lines}  , words:{ words} , characters = {characters}')    

```
