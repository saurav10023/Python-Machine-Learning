#LAMBDA FUNCTION
1.lambda functions are small anonymous function defined using the ( lambda ) keyword  .
2.they can have any number of arguments but only one expression .
3.they are  commonly used for short operations or as arguments to higher - ordeer functions   

```python
addition = lambda a,b : a+b
print(type(addition))
print(addition(1,2))
```

```python
def enev(num):
    if(num%2==0):
        return True
    else : return False
enev(5)    
```

```python
even1 = lambda num : num%2 == 0
even1(2)
```

```python
#map()- applies function to each  elements list or any collections
numbers = [1,2,3,4,5]
def square(number):
    return number**2
square(2)
print(numbers)

```

#map ( function , argument)

```python
list(map(lambda x : x**2 , numbers))
```

The Map() function in python

the map () function applies a given function to all the items in AN INPUT LIST (OR ANY OTHER ITERABLES) and rfeturns a map object (an iterator ) . This is particularly useful for transforming dat in an list comprehensively 


```python
## Map function with iterables
number = [5,6,7,8]
list(map(lambda x : x**2 , number))
```

```python
#Map function with multiple iterables

num1 =[1,2,3]
num2 = [4,5,6]

added_num = list(map(lambda x,y : x+y , num1,num2))
print(added_num)
```

```python
## map () to convert a list to a string to integer
#use map to convert

str_num = ['1', '2' ,'3' , '4' , '5']
int_num = list(map(int , str_num))
print (int_num)
```

```python
# inbuilt functions using map

words = ['apple' , ' banana' , 'cherry']
upper_words = list(map(str.upper , words))
print(upper_words)
```

## Filter() function in python

this is used to filter out items from list or other iterables based on the given condition

```python
lst = [1,2,3,4,5,6,7,8,9]
print(list(filter( lambda x : x>5 and x%2==0, lst)))
```

```python
people = [
    {'name': 'krish' , 'age' :32} ,
    { 'name' : 'jack' , 'age' :33},
    { 'name': 'john' , 'age' : 25}
]

def age_greater_than_30(person):
    return person['age']>25

list(filter(age_greater_than_30 , people))
```
